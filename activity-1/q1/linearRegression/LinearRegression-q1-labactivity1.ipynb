```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import LogisticRegression
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
data=pd.read_csv('taxi.csv')
print(data.head())

```

```python
data.shape
```

```python
data = data.dropna(subset=['trip_seconds', 'trip_miles', 'fare'])
```

```python
data['Payment_Card'] = data['payment_type'].map(lambda p: 1 if p == 'Credit Card' else 0)
```

```python
data.head()
```

```python
x = data[['trip_miles', 'trip_seconds', 'Payment_Card']]
y=data[['fare']]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
model=LinearRegression()
model.fit(x_train,y_train)
predict=model.predict(x_test)
print(predict)
```

```python
print("actual vs prediction weights")

comparison=pd.DataFrame({"actual prediction":y_test.values.flatten(),
                         "prediction":predict.flatten()})
print(comparison)
```

```python
mae=mean_absolute_error(y_test,predict)
mse=mean_squared_error(y_test,predict)
rsquared=r2_score(y_test,predict)
print(mae)
print(mse)
print(rsquared)
```
